````python
import pandas as pd
import numpy as np
````

````python
random_age = np.random.randint(18,50,18)
random_salary = np.random.randint(1500,45000,18)

df = pd.DataFrame({
    'Name' : ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Gerald', 'Harold', 'Ingrid', 'Julian', 'Keith', 'Larry', 'Michael', 'Nicole', 'Peter', 'Oscar', 'Ronald', 'Sarah'],
    'Age' : random_age,
    'Gender' : ['Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'Salary' :  ['1,500','2200','3000+','4500','5,750','7200+','8,900',None,'10500','12000','15,000+','18500','22000',None,'28,750','35000+','40500','45,000']
})
````

````python
print('Dimensiones del dataset',df.shape)
print('Columnas del dataset')
print(list(df.columns))
print('\nTipo de datos del dataset:')
print(df.dtypes)
print('\nPrimeros 5 registros del dataset')
print(df.head())
````

````python
col_str = df['Salary'].value_counts().to_string()
print()
print(col_str)
````

````python
print('Busqueda de Salarios con errores de normalizacion')

#(r'\+|,'), na=False, esto lo que hace es buscar dentro de la columna los valores que contengan '+' o ',' y que se salta los valores nulos.
print(
    df[df['Salary'].str.contains(r'\+|,', na=False)][
        ['Name','Age','Gender','Salary']
      ]
)
````

````python
#Trabajaremos con la columna Salary para reemplazar los valores que tengan '+' o ',' para asi poder convertir la columna a numerica.

df['Salary_Numeric'] = df['Salary'].str.replace(r'\+|,', '', regex=True)
df['Salary_Numeric'] = pd.to_numeric(df['Salary_Numeric'], errors="coerce")
print()
print(df['Salary_Numeric'].value_counts().to_string())
print('\nNuevo tipo de dato para Salary')
print(df['Salary_Numeric'].dtypes)
print('\nValores Faltantes en Salario')
print(df['Salary_Numeric'].isna().sum())
````

````python
print('\nNueva Dimension del dataset', df.shape)
print('Columnas del dataset')
print(list(df.columns))
print('\nTipo de datos del dataset:')
print(df.dtypes)
print('\nPrimeros 5 registros del dataset')
print(df.head())
````

````python
df['Salary_Numeric'] = df['Salary_Numeric'].fillna(df['Salary_Numeric'].mean())
print('Valores faltantes DESPUES de los cambios')
print(df.isnull().sum())
````

# Ejercicio de limpieza y normalización de datos con Pandas

## Objetivo del ejercicio

El objetivo de este ejercicio es crear un pequeño conjunto de datos que simule un problema real de Machine Learning.

La columna **Salary** contiene valores almacenados como texto y presenta problemas de normalización, como:

- Separadores de miles (`,`).
- Símbolos adicionales (`+`).
- Valores faltantes (`None`).

El objetivo es identificar estos problemas, limpiar la información y convertir la columna a un formato numérico.

---

# Búsqueda de errores de normalización

```python
df['Salary'].str.contains(r'\+|,', na=False)
```

`str.contains()` permite buscar patrones dentro de una columna de texto.

La expresión regular:

```python
r'\+|,'
```

significa:

- `\+` → buscar el símbolo `+`.
- `|` → operador lógico "o".
- `,` → buscar una coma.

---

```python
na=False
```

Este parámetro evita errores al encontrar valores nulos (`None`).

---

```python
print(
    df[df['Salary'].str.contains(r'\+|,', na=False)][
        ['Name', 'Age', 'Gender', 'Salary']
    ]

```

# Limpieza de la columna Salary

## Eliminar caracteres especiales

```python
df['Salary_Numeric'] = df['Salary'].str.replace(
    r'\+|,',
    '',
    regex=True
)
```

`str.replace()` elimina todos los símbolos `+` y `,`.

Ejemplos:

| Valor original | Valor limpio |
| --- | --- |
| `1,500` | `1500` |
| `3,000+` | `3000` |
| `15,000+` | `15000` |
| `45,000` | `45000` |

---

## Conversión a formato numérico

```python
df['Salary_Numeric'] = pd.to_numeric(
    df['Salary_Numeric'],
    errors='coerce'
)
```

`pd.to_numeric()` convierte los valores a formato numérico.

```python
errors='coerce'
```

Este parámetro indica que los valores que no puedan convertirse deben reemplazarse por `NaN`.

Ejemplo:

| Valor original | Resultado |
| --- | --- |
| `1500` | `1500` |
| `3000` | `3000` |
| `None` | `NaN` |
