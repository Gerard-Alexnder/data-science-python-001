# Ejercicio de Python aplicado a la limpieza de datos para Machine Learning

## Objetivo

Desarrollar una función personalizada en Python capaz de limpiar automáticamente una columna con valores numéricos almacenados como texto.

El objetivo del ejercicio es reforzar los conceptos fundamentales de Python aplicados al preprocesamiento de datos, simulando situaciones reales que suelen encontrarse en proyectos de Machine Learning.

## Problema

La columna contiene valores inconsistentes, entre ellos:

- Separadores de miles (`,`).
- Símbolos adicionales (`+`).
- Símbolos monetarios (`$`).
- Operadores matemáticos (`<`).
- Valores categóricos (`Electric`, `Hybrid`).
- Valores desconocidos (`-`).
- Valores faltantes (`None`).

Estos registros impiden la conversión directa de la columna a un formato numérico.

## Objetivos específicos

- Crear una función utilizando `for`.
- Aplicar estructuras condicionales (`if`, `elif` y `else`).
- Utilizar el método `replace()` para eliminar caracteres especiales.
- Identificar valores categóricos.
- Reemplazar valores no numéricos por `np.nan`.
- Convertir los valores válidos a un formato numérico.
- Construir una nueva columna limpia utilizando `append()`.

## Dataset del ejercicio

```python
columna_str = ['1,500', '3,000+', '$5,700', '< 1.9', 'Electric', 'Hybrid', '-', None]

````python
import numpy as np
````

````python
def limpiar_columna_numerica(columna):

    columna_numerica = []
    for registro in columna:

        #1. Ignorar valores faltantes
        if registro is None:
            columna_numerica.append(np.nan)
            continue
        
        #2. Convertir temporalmente a string

        registro =str(registro)

        #3. Limpiar Caracteres especiales
        registro = registro.replace(',', '')
        registro = registro.replace('+', '')
        registro = registro.replace('$', '')
        registro = registro.replace('<','')
        registro = registro.strip()
        
        #4. Convertir el valor limpio a numero
        try:

            registro = float(registro)
            columna_numerica.append(registro)

        except ValueError:
            columna_numerica.append(np.nan)
            
    
    return columna_numerica

columna_str = ['1,500','3,000+','$5,700', '< 1.9', 'Electric', 'Hybrid', '-',  None,'Combustion']

print("\nFuncion que limpia una columna automaticamente")
print(f'Columna_Original: {columna_str}')
print(f'Nueva Columna "Columna_Numerica": {limpiar_columna_numerica(columna_str)}')
````

# Explicación del código: función `limpiar_columna_numerica()`

## Objetivo de la función

El objetivo de esta función es limpiar automáticamente una columna que contiene valores numéricos almacenados como texto y convertirlos en datos que puedan utilizarse posteriormente en un proceso de análisis o en un modelo de Machine Learning.

La función identifica y corrige diferentes problemas presentes en los datos, como:

- Separadores de miles (`,`).
- Símbolos adicionales (`+`).
- Símbolos monetarios (`$`).
- Operadores matemáticos (`<`).
- Valores categóricos (`Electric`, `Hybrid`, `Combustion`, etc.).
- Valores faltantes (`None`).

---

## Importación de NumPy

```python
import numpy as np
```

Se importa la librería **NumPy** para utilizar `np.nan`, que representa un valor faltante dentro de una columna numérica.

---

## Definición de la función

```python
def limpiar_columna_numerica(columna):
```

La función recibe como parámetro una columna o una lista con los datos que se desean limpiar.

---

## Creación de una nueva lista

```python
columna_numerica = []
```

Se crea una lista vacía donde se almacenarán los valores procesados.

La idea es conservar los registros válidos y reemplazar los valores incorrectos por `np.nan`.

---

## Recorrer cada registro

```python
for registro in columna:
```

El bucle `for` recorre cada elemento de la columna de manera individual.

Por ejemplo:

```text
'1,500'
'3,000+'
'$5,700'
'< 1.9'
'Electric'
'Hybrid'
'-'
None
```

---

## Paso 1: identificar valores faltantes

```python
if registro is None:
    columna_numerica.append(np.nan)
    continue
```

Si el registro contiene `None`, se reemplaza por `np.nan`.

El comando `continue` permite continuar con el siguiente elemento sin ejecutar el resto del código.

---

## Paso 2: convertir temporalmente el registro en una cadena de texto

```python
registro = str(registro)
```

Todos los registros se convierten temporalmente a `string` para poder utilizar métodos como `replace()`.

---

## Paso 3: eliminar caracteres especiales

```python
registro = registro.replace(',', '')
registro = registro.replace('+', '')
registro = registro.replace('$', '')
registro = registro.replace('<', '')
registro = registro.strip()
```

Se eliminan los caracteres que impiden convertir el valor a un formato numérico.

Ejemplos:

| Valor original | Valor limpio |
| --- | --- |
| `1,500` | `1500` |
| `3,000+` | `3000` |
| `$5,700` | `5700` |
| `< 1.9` | `1.9` |

El método `strip()` elimina los espacios en blanco al principio y al final del texto.

---

## Paso 4: convertir el valor limpio a un número

```python
try:

    registro = float(registro)
    columna_numerica.append(registro)
```

La estructura `try` intenta convertir el registro a un número decimal utilizando `float()`.

Si la conversión es exitosa, el valor se almacena en la nueva lista.

---

## Manejo de errores con `except`

```python
except ValueError:
    columna_numerica.append(np.nan)
```

Si el registro no puede convertirse a un número, Python genera una excepción (`ValueError`).

En ese caso, el valor se reemplaza por `np.nan`.

Ejemplos:

| Valor | Resultado |
| --- | --- |
| `Electric` | `np.nan` |
| `Hybrid` | `np.nan` |
| `Combustion` | `np.nan` |
| `-` | `np.nan` |

---

## Retornar la nueva columna

```python
return columna_numerica
```

La función devuelve una nueva lista con los datos ya procesados.
