### Importar Librerias

```python
import pandas as pd
import numpy as np
```

### Carga de datasets y Exploracion inicial

```python
df = pd.read_csv('../datasets/NY-House-Dataset.csv')
```

```python
print("--Exploracion Inicial del Dataset--")
print('Dimensiones del Dataset:', df.shape)
print('\nColumnas del Dataset')
print(list(df.columns))
print('\nTipos de datos de las columnas del dataset')
print(df.dtypes)
print('\nPrimeras filas del dataset')
print(df.head())
```

### Busqueda de valores faltantes y Duplicados

```python
print("Busqueda de valores faltantes")
print(df.isnull().sum())
print('\nBusqueda de datos duplicados', df.duplicated().sum())
```

```python
#Duplicados completos
duplicados_completo = df[df.duplicated(keep=False)]

print("Filas Involucradas en duplicados")
print(len(duplicados_completo))
print("\nNumero de duplicados addicionales")
print(df.duplicated().sum())
print("\nDuplicados Encontrados:")
print(duplicados_completo.sort_values('FORMATTED_ADDRESS').head(20).to_string())

```

```python
#Eliminar los duplicados encontrados
df = df.drop_duplicates().reset_index(drop=True)
print("Duplicados restantes:", df.duplicated().sum())
print("Nuevas dimensiones del dataset:", df.shape)
```

### Análisis y tratamiento de registros 

Durante la exploración inicial se identificaron **0** valores faltantes, pero hubieron **214 registros duplicados** mediante `duplicated().sum()`.

Antes de eliminarlos, se realizó una inspección utilizando `duplicated(keep=False)` para visualizar tanto los registros originales como sus respectivas copias. Se revisaron diferentes muestras de los registros duplicados y se comprobó que las filas contenían los mismos valores en todas las variables, incluyendo precio y características de la propiedad.

Al tratarse de duplicados completos y no de propiedades con características o precios diferentes, se decidió eliminar las copias mediante `drop_duplicates()`.

Finalmente, se restableció el índice del DataFrame con `reset_index(drop=True)` y se verificó nuevamente la cantidad de duplicados para confirmar que el dataset no conservara registros repetidos.

```python
columnas_categoricas = df.select_dtypes(include='str').columns
columnas_numericas = df.select_dtypes(include=['int','float']).columns

print("Columnas Categoricas del dataset:")
print(columnas_categoricas)
print('\nColumnas Numericas del datset:')
print(columnas_numericas)
```
